```python
"""Tabella riassuntiva delle metriche di ricostruzione (RMSE / NMAE) per modello.

Colonne: RMSE e NMAE aggregati (Unify all) + NMAE per singolo campo (SE, Uvel, Vvel),
ciascuno con Train / Val / Test, piu' la barra di miglioramento sul test NMAE.
"""
import json

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import Rectangle

# --------------------------------------------------------------------------- #
# Configurazione
# --------------------------------------------------------------------------- #
METRICS_PATH = "1-reconstruction_metrics.json"
OUT_STEM = "HYDRO_3_table"
SPLITS = ("train", "val", "test")
FONT = 17

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "DejaVu Serif"],
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

C_BASE, C_IMPUT, C_MASK, C_GRUD = "#000000", "#1F4E9C", "#B8860B", "#7F7F7F"
C_BAR = "#2ca02c"

GROUPS = {
    "0": ("B", "Baselines", C_BASE),
    "1": ("I", "Imputation", C_IMPUT),
    "2": ("G", "GRU-D", C_GRUD),
    "3": ("M", "Masking", C_MASK),
}
MODEL_RENAME = {
    "T0A": "Always-available sensors + SHRED",
    "T0B": "Always-available sensors + T-SHRED",
    "T1A": "Median imputation + T-SHRED",
    "T1B": "Mean imputation + T-SHRED",
    "T1C": "LOCF + T-SHRED",
    "T2A": "GRUD",
    "T2B": "Attn-GRUD",
    "T3A": "Input masking",
    "T3B": "Channel-wise masking",
    "T3C": "Hidden-space masking",
    "T3D": "MHSA masking",
    "T3E": "Channel-wise and MHSA masking",
    "T3F": "Learned MHSA masking",
    "T3G": "Channel-wise and learned MHSA masking",
}

# gruppi di colonne numeriche: (titolo sotto-intestazione, prefisso colonna df)
# colonna df = f"{split}-{suffix}"
NUM_GROUPS = [("RMSE", "rmse"), ("NMAE", "nmae")]


# --------------------------------------------------------------------------- #
# Dati
# --------------------------------------------------------------------------- #
def load_dataframe(path):
    """Legge il JSON e restituisce un DataFrame ordinato per test NMAE (unify all)."""
    with open(path, encoding="utf-8") as f:
        data = json.load(f)

    rows = []
    for model, splits in data.items():
        row = {"model": model, "group": model[1]}
        for s in SPLITS:
            agg = splits[s]["Unify all"]
            row[f"{s}-rmse"] = np.sqrt(agg["mse"])
            row[f"{s}-nmae"] = agg["nmae"]
        rows.append(row)

    df = pd.DataFrame(rows).sort_values("test-nmae", kind="stable", ignore_index=True)
    df["name"] = df["model"].map(MODEL_RENAME).fillna(df["model"])
    worst = df["test-nmae"].max()
    df["impr"] = (worst - df["test-nmae"]) / worst * 100
    return df


# --------------------------------------------------------------------------- #
# Utilita' di layout
# --------------------------------------------------------------------------- #
def text_width_in(texts, fontsize=FONT, weight="normal"):
    fig = plt.figure()
    renderer = fig.canvas.get_renderer()
    w = max(
        fig.text(0, 0, t, fontsize=fontsize, fontweight=weight)
        .get_window_extent(renderer).width
        for t in texts
    ) / fig.dpi
    plt.close(fig)
    return w


# --------------------------------------------------------------------------- #
# Disegno
# --------------------------------------------------------------------------- #
def draw_table(df):
    best_impr = df["impr"].max()

    pad = 0.09
    first_col_w = text_width_in(list(df["name"]) + ["Configuration"], weight="bold") + 2 * pad + 0.1
    num_col_w = 0.92 * 1.10
    gap_w = 0.25

    # costruzione delle colonne: [type, config, (3 num + gap) * n_gruppi, barra]
    widths = [0.8, first_col_w]
    group_cols = []                      # lista di liste di indici (train, val, test)
    for _ in NUM_GROUPS:
        start = len(widths)
        widths += [num_col_w] * 3
        group_cols.append([start, start + 1, start + 2])
        widths.append(gap_w)
    widths.append(3.4 * 1.30)
    BAR = len(widths) - 1

    xs = np.concatenate([[0], np.cumsum(widths)[:-1]])
    total_w = float(sum(widths))

    # colonne df corrispondenti a ciascun indice numerico
    col_map = {}
    for (_, suffix), idxs in zip(NUM_GROUPS, group_cols):
        for c, s in zip(idxs, SPLITS):
            col_map[c] = f"{s}-{suffix}"
    best = df[list(col_map.values())].min()

    row_h = 0.28 * FONT / 11
    n_rows, header_rows, legend_rows = len(df), 2, 1.0
    fig_h = (n_rows + header_rows + legend_rows + 0.2) * row_h + 0.1
    fig = plt.figure(figsize=(total_w, fig_h))
    ax = fig.add_axes([0, 0, 1, 1])
    ax.set_xlim(0, total_w)
    ax.set_ylim(n_rows + header_rows + legend_rows + 0.1, -0.1)
    ax.axis("off")

    def put(col, row, text, ha="left", **kw):
        x = {"left": xs[col] + pad,
             "right": xs[col] + widths[col] - pad,
             "center": xs[col] + widths[col] / 2}[ha]
        ax.text(x, row + 0.5, text, ha=ha, va="center", fontsize=FONT, **kw)

    def hline(x0, x1, y, lw):
        ax.plot([x0, x1], [y, y], color="black", lw=lw)

    def span(c0, c1, row, label):
        """Etichetta centrata sulle colonne c0..c1 con sottolineatura."""
        x0, x1 = xs[c0], xs[c1] + widths[c1]
        ax.text((x0 + x1) / 2, row + 0.5, label, ha="center", va="center",
                fontsize=FONT, fontweight="bold")
        hline(x0 + 0.05, x1 - 0.05, row + 1, 0.6)

    # ---- intestazione ----
    # riga 0: titoli dei gruppi di colonne
    for (title, _), idxs in zip(NUM_GROUPS, group_cols):
        span(idxs[0], idxs[-1], 0, title)
    span(BAR, BAR, 0, "Improvement vs. Baseline")
    # riga 1: split
    ax.text(xs[BAR] + widths[BAR] / 2, 1.5, "Test NMAE (%)", ha="center",
            va="center", fontsize=FONT, fontweight="bold")
    put(0, 1, "Type", ha="center", fontweight="bold")
    put(1, 1, "Configuration", fontweight="bold")
    for idxs in group_cols:
        for c, name in zip(idxs, ["Train", "Val", "Test"]):
            put(c, 1, name, ha="right", fontweight="bold")

    # ---- scala delle barre ----
    label_w = 1.0
    bx0 = xs[BAR] + 0.15
    usable = widths[BAR] - 0.15 - label_w - 0.1
    vmin, vmax = min(df["impr"].min(), 0.0), max(best_impr, 1e-9)
    scale = usable / (vmax - vmin)
    zero_x = bx0 - vmin * scale
    bar_h = 0.62

    # ---- corpo ----
    for r, row in df.iterrows():
        y = r + header_rows
        letter, _, color = GROUPS.get(str(row["group"]), ("?", "", "black"))
        put(0, y, letter, ha="center", color=color, fontweight="bold")
        put(1, y, row["name"])
        for c, metric in col_map.items():
            put(c, y, f"{row[metric]:.4f}", ha="right",
                fontweight="bold" if row[metric] == best[metric] else "normal")

        v = row["impr"]
        w = v * scale
        ax.add_patch(Rectangle((min(zero_x, zero_x + w), y + 0.5 - bar_h / 2),
                               abs(w), bar_h, facecolor=C_BAR, edgecolor="none"))
        sign = 1 if v >= 0 else -1
        ax.text(zero_x + w + sign * 0.06, y + 0.5, f"{v:.2f}%",
                ha="left" if v >= 0 else "right", va="center", fontsize=FONT - 1,
                fontweight="bold" if np.isclose(v, best_impr) else "normal")

    ax.plot([zero_x, zero_x], [header_rows, n_rows + header_rows], color="black", lw=0.6)
    hline(0, total_w, 0, 1.4)
    hline(0, total_w, header_rows, 0.8)
    hline(0, total_w, n_rows + header_rows, 1.4)

    # ---- legenda ----
    ly = n_rows + header_rows + legend_rows / 2
    sq, sq_h, entry_w = 0.17, 0.17 / row_h, 2.3
    x_start = (total_w - entry_w * len(GROUPS)) / 2
    for i, (letter, label, color) in enumerate(GROUPS.values()):
        x = x_start + i * entry_w
        ax.add_patch(Rectangle((x, ly - sq_h / 2), sq, sq_h, facecolor=color, edgecolor="none"))
        ax.text(x + sq + 0.12, ly, f"{letter}: {label}", ha="left", va="center", fontsize=FONT)

    return fig


def main():
    df = load_dataframe(METRICS_PATH)
    fig = draw_table(df)
    fig.savefig(f"{OUT_STEM}.png", dpi=500, bbox_inches="tight", pad_inches=0.03)
    plt.show()


if __name__ == "__main__":
    main()
```

```python
import json
import math
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# =====================================================================
# CONFIGURAZIONE
# =====================================================================

FILE_CANDIDATES = [
    Path("2-reconstruction_metrics.json"),
]

INPUT_FILE = next((p for p in FILE_CANDIDATES if p.is_file()), None)

if INPUT_FILE is None:
    raise FileNotFoundError(
        "File non trovato. Atteso '2-reconstruction' "
        "oppure '2-reconstruction.json' nella directory corrente."
    )

OUT = "HYDRO_4_lagged_training"

# Variabile del JSON usata per le metriche (blocco aggregato).
VARIABLE = "Unify all"

# Mostra o nasconde la riga con il nome del modello.
SHOW_MODEL_TITLE = False

MODEL_NAMES = {
    "T3G": "Channel-wise and learned MHSA masking",
}

TRAINING_NAMES = {
    "last_step": "Last-step training",
    "multi_step": "Multi-step training",
}

METRICS = ["rmse", "nmae"]
SPLITS = ["train", "val", "test"]

RMSE_DEC = 5
NMAE_DEC = 4

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "Times", "STIXGeneral", "DejaVu Serif"],
    "mathtext.fontset": "stix",
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

S = 1.5
FS = 7.5 * S
GAP = 0.35

C_BAR = "#2ca02c"

# =====================================================================
# CARICAMENTO JSON
# =====================================================================

with INPUT_FILE.open("r", encoding="utf-8") as f:
    raw_data = json.load(f)


def get_metric(metric, split, model_key, variable=VARIABLE):
    """RMSE = sqrt(MSE); NMAE = valore nmae presente nel JSON."""
    values = raw_data[model_key][split][variable]

    if metric == "rmse":
        mse = values.get("mse")
        return math.sqrt(mse) if mse is not None else float("nan")

    if metric == "nmae":
        nmae = values.get("nmae")
        return float(nmae) if nmae is not None else float("nan")

    raise ValueError(f"Metrica non supportata: {metric}")


# =====================================================================
# COSTRUZIONE DELLA TABELLA
# =====================================================================

rows = []

for model_id, model_label in MODEL_NAMES.items():

    if rows:
        rows.append(("gap", None))

    if SHOW_MODEL_TITLE:
        rows.append(("title", model_label))

    model_keys = {
        "last_step": f"last_step_{model_id}",
        "multi_step": f"multi_step_{model_id}",
    }

    model_values = {}
    for training_id, json_key in model_keys.items():
        model_values[training_id] = {
            split: {m: get_metric(m, split, json_key) for m in METRICS}
            for split in SPLITS
        }

    best = {}
    for split in SPLITS:
        for metric in METRICS:
            vals = [model_values[t][split][metric] for t in model_keys]
            vals = [v for v in vals if math.isfinite(v)]
            best[(split, metric)] = min(vals) if vals else float("nan")

    baseline_rmse = model_values["last_step"]["test"]["rmse"]

    for training_id, training_label in TRAINING_NAMES.items():
        test_rmse = model_values[training_id]["test"]["rmse"]

        if (
            training_id == "last_step"
            or not math.isfinite(baseline_rmse)
            or baseline_rmse == 0
            or not math.isfinite(test_rmse)
        ):
            improvement = 0.0
        else:
            improvement = (baseline_rmse - test_rmse) / baseline_rmse * 100

        rows.append((
            "data",
            {
                "name": training_id,
                "label": training_label,
                "vals": model_values[training_id],
                "best": best,
                "impr": improvement,
                "is_base": training_id == "last_step",
            },
        ))

positive_improvements = [
    r[1]["impr"] for r in rows if r[0] == "data" and r[1]["impr"] > 0
]
best_impr = max(positive_improvements, default=1.0)

# =====================================================================
# LAYOUT: Model | 3 RMSE | gap | 3 NMAE | gap | Improvement
# =====================================================================

nw = 0.62 * S

widths = [1.65 * S] + [nw] * 3 + [0.15 * S] + [nw] * 3 + [0.15 * S] + [1.95 * S]

xs = [sum(widths[:i]) for i in range(len(widths))]
total_w = sum(widths)

NUM_COLS = [1, 2, 3, 5, 6, 7]
BAR = len(widths) - 1

row_h = 0.20 * S
header_rows = 2

body_h = sum(GAP if kind == "gap" else 1 for kind, _ in rows)
fig_h = (body_h + header_rows) * row_h

fig, ax = plt.subplots(figsize=(total_w, fig_h))
ax.set_xlim(0, total_w)
ax.set_ylim(body_h + header_rows, 0)
ax.axis("off")

pad = 0.05 * S


def put(col, row, text, ha="left", **kw):
    x = {
        "left": xs[col] + pad,
        "right": xs[col] + widths[col] - pad,
        "center": xs[col] + widths[col] / 2,
    }[ha]
    ax.text(x, row + 0.5, text, ha=ha, va="center", fontsize=FS, **kw)


def span(c0, c1):
    return xs[c0] + (xs[c1] + widths[c1] - xs[c0]) / 2


# =====================================================================
# HEADER
# =====================================================================

for c0, c1, title in [(1, 3, "RMSE"), (5, 7, "NMAE")]:
    ax.text(span(c0, c1), 0.5, title, ha="center", va="center",
            fontsize=FS, fontweight="bold")
    ax.plot([xs[c0] + 0.03 * S, xs[c1] + widths[c1] - 0.03 * S], [1, 1],
            color="black", lw=0.5 * S)

ax.text(span(BAR, BAR), 0.5, "Improvement", ha="center", va="center",
        fontsize=FS, fontweight="bold")
ax.plot([xs[BAR] + 0.03 * S, total_w], [1, 1], color="black", lw=0.5 * S)

put(0, 1, "Model", fontweight="bold")

for col, label in zip(NUM_COLS, ["Train", "Val", "Test"] * 2):
    put(col, 1, label, ha="right", fontweight="bold")

ax.text(span(BAR, BAR), 1.5, "Test RMSE vs. Last-step (%)", ha="center",
        va="center", fontsize=FS, fontweight="bold")

# =====================================================================
# BARRE DI IMPROVEMENT
# =====================================================================

label_w = 0.50 * S
zero_x = xs[BAR] + 0.10 * S
scale = (widths[BAR] - 0.10 * S - label_w) / best_impr
bar_h = 0.60

# =====================================================================
# CORPO
# =====================================================================

y = header_rows

for kind, content in rows:

    if kind == "gap":
        ax.plot([0, total_w], [y, y], color="black", lw=0.3 * S)
        y += GAP
        continue

    if kind == "title":
        ax.text(pad, y + 0.5, content, ha="left", va="center",
                fontsize=FS, fontstyle="italic", fontweight="bold")
        y += 1
        continue

    r = content
    put(0, y, r["label"])

    for col, split, metric in zip(
        NUM_COLS,
        ["train", "val", "test"] * 2,
        ["rmse"] * 3 + ["nmae"] * 3,
    ):
        value = r["vals"][split][metric]
        best_value = r["best"][(split, metric)]
        decimals = RMSE_DEC if metric == "rmse" else NMAE_DEC

        text_value = f"{value:.{decimals}f}" if math.isfinite(value) else "--"

        is_best = (
            math.isfinite(value)
            and math.isfinite(best_value)
            and math.isclose(value, best_value, rel_tol=1e-12, abs_tol=1e-15)
        )

        put(col, y, text_value, ha="right",
            fontweight="bold" if is_best else "normal")

    improvement = r["impr"]

    if r["is_base"] or improvement == 0:
        ax.text(zero_x + 0.04 * S, y + 0.5, "0.00%", ha="left",
                va="center", fontsize=FS)
    elif improvement > 0:
        w = improvement * scale
        ax.add_patch(Rectangle((zero_x, y + (1 - bar_h) / 2), w, bar_h,
                               facecolor=C_BAR, edgecolor="none"))
        ax.text(zero_x + w + 0.05 * S, y + 0.5, f"+{improvement:.2f}%",
                ha="left", va="center", fontsize=FS)
    else:
        ax.text(zero_x + 0.04 * S, y + 0.5, f"{improvement:.2f}%",
                ha="left", va="center", fontsize=FS)

    y += 1

ax.plot([zero_x, zero_x], [header_rows, body_h + header_rows],
        color="black", lw=0.5 * S)

# =====================================================================
# REGOLE BOOKTABS
# =====================================================================

ax.plot([0, total_w], [0, 0], color="black", lw=1.0 * S)
ax.plot([0, total_w], [header_rows, header_rows], color="black", lw=0.6 * S)
ax.plot([0, total_w], [body_h + header_rows, body_h + header_rows],
        color="black", lw=1.0 * S)

# =====================================================================
# ESPORTAZIONE
# =====================================================================

fig.savefig(OUT + ".png", dpi=500, bbox_inches="tight", pad_inches=0.02 * S)
```
